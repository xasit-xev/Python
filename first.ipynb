```python
winning_number = 23
number = int(input("Guess the winning number between 1 and 100: "))
if number == winning_number:
    print("Congratulations! You guessed the winning number.")
else:
    if number > winning_number:
        print(f"{number} Sorry, that's not the winning number. Try again! The winning number is higher.")
    else:
        if number < winning_number:
            print(f"{number} Sorry, that's not the winning number. Try again! The winning number is lower.")
```
